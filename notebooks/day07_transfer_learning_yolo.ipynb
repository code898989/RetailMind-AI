# RetailMind AI — Day 7: Transfer Learning & Object Detection
**Objective**: Fine-tune a pretrained ResNet50 backbone on product imagery and run Ultralytics YOLO to detect products and bounding boxes for catalog QA.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.applications import ResNet50
from ultralytics import YOLO

# 1. Load pretrained ResNet50 backbone
base_model = ResNet50(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False
print('ResNet50 backbone loaded with frozen weights.')

# 2. Initialize YOLO detector
detector = YOLO('yolov8n.pt')
sample = np.ones((400, 400, 3), dtype=np.uint8) * 255
results = detector.predict(sample, verbose=False)
print('YOLO detector initialized.')
```
